```python
import pandas as pd

filepath = "data/tiktok.csv"
df = pd.read_csv(filepath)

logger.info("Loaded %s | shape=%s", filepath, df.shape)
print(df.shape)

(df.head(10))
```

# checking missing values

```python
print(df.isna().sum())
print((df.isna().mean()*100).round(1))

logger.info("Missing values | %s", (df.isna().mean()*100).round(1))
```

# convert Reply to which Comment to a boolean signal : is_reply

```python
# --- Reply to Which Comment (53.9% missing) ---
# Mechanism: structural (NaN = top-level comment, not a data gap)
# Decision: don't impute a fake ID — convert to a boolean signal
df["is_reply"] = df["Reply to Which Comment"].notna()
print(df.shape)
print(df.isna().sum())

logger.info("Reply to Which Comment | converted to boolean signal 'is_reply'")
df.head(10)
```

# filling missing values of column Reply Count as 0

```python
# --- Reply Count (46.1% missing) ---
# Mechanism: structural (only top-level comments have a reply count)
# Decision: fill with 0 (a reply logically has 0 replies-to-it, tracked elsewhere)
df["Reply Count"] = df["Reply Count"].fillna(0).astype(int)
print(df.shape)
print(df.isna().sum())
logger.info("Reply Count | filled missing with 0 and converted to int")
df.head(10)
```

# Pinned to Top filled with Not Applicable

```python
df["Pinned to Top"] = df["Pinned to Top"].fillna("Not Applicable")

print(df["Pinned to Top"].value_counts())
print(df.shape)
print(df.isna().sum()) 
logger.info("Pinned to Top | filled missing with 'Not Applicable'")
df.head(10 )
```

# Reply to Which Comment filled with Not Reply to Any Comment

```python
df["Reply to Which Comment"] = df["Reply to Which Comment"].fillna("Not Reply to Any Comment")
logger.info("Reply to Which Comment | filled missing with 'Not Reply to Any Comment'")
print(df.shape)
print(df.isna().sum())
df.head(10 )
```

# Dropping the missing row with no comment

```python

df = df.dropna(subset=["Comment"])
print(df.isna().sum()) 
print(df.shape)
logger.info("rows with missing 'Comment' dropped | shape=%s", df.shape)
df.head(20)
```

# new column is_emoji added to clarify emoji or text in commment

```python

import emoji  # pip install emoji



def classify(comment):
    if pd.isna(comment):
        return None  # can't classify missing text
    has_emoji = bool(emoji.emoji_list(comment))
    text_only = emoji.replace_emoji(comment, replace="").strip()
    has_text = len(text_only) > 0

    if has_emoji and has_text:
        return "text"      # mixed → counted as text (your rule)
    elif has_emoji:
        return "emoji"     # emoji-only
    else:
        return "text"      # plain text, no emoji

df["is_emoji"] = df["Comment"].apply(classify)

print(df["is_emoji"].value_counts(dropna=False))
print(df.shape)
print(df.isna().sum())
df.head(15)
```

# save clean data in tiktok_clean.csv

```python
df_clean = df.copy()
df_clean.to_csv("data/tiktok_clean.csv", index=False)
logger.info("Cleaned data saved to 'data/tiktok_clean.csv' Shape=%s  %s" , df_clean.shape, df_clean.isna().sum())
df_clean.head(10)
```

# Beautiful Soup example

```python
from bs4 import BeautifulSoup 
html = """
<html>
<head><title>My Test Page</title></head>
<body>
    <h1>Welcome to Web Scraping</h1>
    <div class="comment">
        <p class="author">Nawaraj</p>
        <p class="text">This is a great tutorial!</p>
    </div>
    <div class="comment">
        <p class="author">Aashish</p>
        <p class="text">Thanks for sharing.</p>
    </div>
    <a href="https://example.com">Visit Example</a>
</body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")

print("Title:", soup.title.text)
print("H1 text:", soup.h1.text)

print("\nAll authors:")
for author in soup.find_all("p", class_="author"):
    print(" -", author.text)

print("\nAll comment texts:")
for text in soup.find_all("p", class_="text"):
    print(" -", text.text)

print("\nLink href:", soup.a["href"])
```

# with a real webpage instead of a fixed string

```python
import requests
from bs4 import BeautifulSoup

url = "https://quotes.toscrape.com/"
response = requests.get(url)

print("Status code:", response.status_code)

soup = BeautifulSoup(response.text, "html.parser")

# Each quote is inside a <div class="quote">
quotes = soup.find_all("div", class_="quote")
print(f"Found {len(quotes)} quotes on this page\n")

for q in quotes:
    text = q.find("span", class_="text").text
    author = q.find("small", class_="author").text
    tags = [tag.text for tag in q.find_all("a", class_="tag")]
    print(f"{text}\n— {author}")
    print("Tags:", tags)
    print("-" * 60)
```

```python
from selenium import webdriver
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.firefox import GeckoDriverManager

service = Service(GeckoDriverManager().install())
driver = webdriver.Firefox(service=service)

driver.get("https://quotes.toscrape.com/js/")

quotes = driver.find_elements(By.CLASS_NAME, "quote")
print(f"Found {len(quotes)} quotes\n")

for q in quotes[:3]:
    text = q.find_element(By.CLASS_NAME, "text").text
    author = q.find_element(By.CLASS_NAME, "author").text
    print(f"{text}\n— {author}\n")

driver.quit()
```
